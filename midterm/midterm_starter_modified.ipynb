```python
# Cell A1 - Imports
# Run this cell first. If any import fails, check your conda environment.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import warnings
warnings.filterwarnings('ignore')

# scikit-learn
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Matminer — dataset loader and featurizers
from matminer.datasets import load_dataset
from matminer.featurizers.composition import ElementProperty
from matminer.featurizers.conversions import StructureToComposition
from matminer.featurizers.base import MultipleFeaturizer
from pymatgen.core import Composition

print('✅ All imports successful')
print('   No MP API key needed for Part A - matbench_mp_gap is a static, versioned dataset.')
```

```python
# Cell A2 — Load the matbench_mp_gap dataset
# This is a modern, standardized benchmark dataset for reproducing Ward et al.'s
# approach. Ward et al. used the OQMD; matbench_mp_gap uses the Materials Project -
# both are large DFT databases of inorganic compounds. The dataset contains
# 106,113 compounds with DFT-PBE band gaps and is hosted via matminer.
#
# First run: downloads and caches the dataset locally (~30-60 seconds, ~depends on connection).
# Subsequent runs: loads instantly from the local cache.
#
# Columns (confirmed from the matminer dataset registry):
#   structure  - Pymatgen Structure object (full crystal structure, not just composition)
#   gap pbe    - target variable: DFT-PBE band gap in eV

print('Loading matbench_mp_gap (106,113 compounds)...')
print('(First run downloads ~100KB compressed; may take 30-60s)')

df = load_dataset('matbench_mp_gap')

print(f'\n✅ Dataset loaded')
print(f'   Shape: {df.shape}')
print(f'   Columns: {df.columns.tolist()}')
print(f'   Band gap range: {df["gap pbe"].min():.3f} - {df["gap pbe"].max():.3f} eV')
print(f'   Mean band gap:  {df["gap pbe"].mean():.3f} eV')
```

```python
# Cell A3 - Build Composition objects and a CSV-friendly subset
# Magpie features require a Pymatgen Composition. We derive it from each Structure's
# formula. We also pull crystal_system directly from the structure here. You will
# need it later for Extension Option 2 (prototype-aware splitting).
#
# We work with a 20,000-entry random subsample of the full 106,113 for this assignment.
# This keeps featurization time manageable (~2-4 min instead of ~20-40 min) while still
# giving you a dataset large enough to draw real conclusions. You may use the full
# dataset if you have the compute time. It is not required and will not earn extra credit.
#
# Note: you may see repeated "spglib: ssm_get_exact_positions failed" messages below.
# This is expected and harmless, spglib (the symmetry library pymatgen uses internally)
# is failing to refine a small number of structures to its tightest numerical tolerance,
# but it still falls back to a valid (just slightly coarser) symmetry determination.
# These messages come from spglib's C-level logger, not Python, so they print regardless
# of warnings.filterwarnings(); we redirect them below to keep the output clean.

import os, sys

SUBSET_SIZE = 20_000
df = df.sample(n=SUBSET_SIZE, random_state=42).reset_index(drop=True)

df['composition'] = df['structure'].apply(lambda s: s.composition)
df['formula'] = df['composition'].apply(lambda c: c.reduced_formula)

devnull_fd = os.open(os.devnull, os.O_WRONLY)
saved_stderr_fd = os.dup(2)
os.dup2(devnull_fd, 2)
try:
    df['crystal_system'] = df['structure'].apply(
        lambda s: s.get_space_group_info()[0] if s is not None else None
    )
finally:
    os.dup2(saved_stderr_fd, 2)
    os.close(devnull_fd)
    os.close(saved_stderr_fd)

df['band_gap'] = df['gap pbe']   # renamed for readability in the rest of the notebook
df['nsites'] = df['structure'].apply(lambda s: len(s))

n_missing_cs = df['crystal_system'].isna().sum()
print(f'Working subset: {df.shape[0]:,} compounds (sampled from 106,113 total)')
print(f'Band gap range: {df.band_gap.min():.3f} - {df.band_gap.max():.3f} eV')
print(f'Mean band gap:  {df.band_gap.mean():.3f} eV')
print(f'Structures with no crystal_system found: {n_missing_cs} '
      f'({100*n_missing_cs/len(df):.2f}% - these will be dropped in Cell A4)')
print()
df[['formula', 'band_gap', 'crystal_system', 'nsites']].head()
```

```python
# Cell A4 - Featurize with Magpie descriptors (Ward et al. descriptor set)
# The Magpie descriptor set uses statistics of elemental properties
# (mean, range, min, max, mode, variance) computed over all elements
# present in the composition. This is a composition-only descriptor:
# it does not use any structural information - Ward et al.'s original approach.
# (Extension Option 3 adds structural descriptors on top of this later.)

print('Generating Magpie composition descriptors...')
print('(This may take 1–3 minutes for 20,000 compounds)')

featurizer = ElementProperty.from_preset('magpie')
featurizer.featurize_dataframe(df, col_id='composition', ignore_errors=True, inplace=True)

# Identify feature columns (all Magpie descriptor columns)
feature_cols = featurizer.feature_labels()

# Drop rows with any NaN features
df_clean = df.dropna(subset=feature_cols + ['band_gap']).reset_index(drop=True)

print(f'\n✅ Featurization complete')
print(f'   Entries before cleaning: {len(df)}')
print(f'   Entries after cleaning:  {len(df_clean)}')
print(f'   Number of Magpie features: {len(feature_cols)}')
print(f'\nFirst 5 feature names: {feature_cols[:5]}')
```

```python
# Cell A5 - Save featurized dataset to CSV (avoids re-fetching/re-featurizing if notebook restarts)
# Note: the Structure objects themselves are NOT saved to CSV (they don't serialize cleanly) -
# only the scalar columns and Magpie features needed for Parts B and C. If Extension Option 2
# or 3 needs the original Structure objects again later, re-run Cells A2-A4.

CACHE_FILE = 'midterm_dataset.csv'

cols_to_save = ['formula', 'band_gap', 'crystal_system', 'nsites'] + feature_cols
df_clean[cols_to_save].to_csv(CACHE_FILE, index=False)

print(f'✅ Dataset saved to {CACHE_FILE}')
print(f'   Shape: {df_clean[cols_to_save].shape}')
print()
print('If you restart this notebook and only need the cached scalar/feature columns:')
print("  df_clean = pd.read_csv('midterm_dataset.csv')")
print("  feature_cols = [c for c in df_clean.columns if 'MagpieData' in c]")
print('  # If you need the Structure objects again (e.g. for Extension 2 or 3),')
print('  # re-run Cells A2-A4 instead - Structures are not cached to CSV.')
```

```python
# Cell A6 - Quick EDA on the dataset

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# Bandgap distribution
axes[0].hist(df_clean['band_gap'], bins=60, color='#2E75B6', edgecolor='white', linewidth=0.4)
axes[0].set_xlabel('Band Gap (eV)', fontsize=11)
axes[0].set_ylabel('Count', fontsize=11)
axes[0].set_title('Bandgap distribution', fontsize=12)
axes[0].axvline(df_clean['band_gap'].mean(), color='#E24B4A', linestyle='--',
                label=f'Mean = {df_clean["band_gap"].mean():.2f} eV')
axes[0].legend(fontsize=9)

# Crystal system composition
crystal_counts = df_clean['crystal_system'].value_counts()
axes[1].barh(crystal_counts.index, crystal_counts.values, color='#1F3864')
axes[1].set_xlabel('Count', fontsize=11)
axes[1].set_title('Crystal system breakdown', fontsize=12)

# Bandgap vs. number of sites (formation energy is not in matbench_mp_gap,
# so we substitute nsites - unit cell size - as the third EDA panel)
sc = axes[2].scatter(
    df_clean['nsites'],
    df_clean['band_gap'],
    alpha=0.4, s=8, c=df_clean['nsites'],
    cmap='viridis'
)
plt.colorbar(sc, ax=axes[2], label='# sites')
axes[2].set_xlabel('Number of sites in unit cell', fontsize=11)
axes[2].set_ylabel('Band gap (eV)', fontsize=11)
axes[2].set_title('Bandgap vs. unit cell size', fontsize=12)

plt.suptitle(f'Dataset overview: {len(df_clean)} compounds', fontsize=13, y=1.02)
plt.tight_layout()
plt.savefig('midterm_eda.png', dpi=150, bbox_inches='tight')
plt.show()
print('EDA plot saved as midterm_eda.png')
```
